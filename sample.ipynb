```python

import pandas as pd
from pathlib import Path

file_path = Path("data") / "SBALoanData-LoanApproval_DefaultAnalysis.csv"

df = pd.read_csv(file_path)

print("Dataset loaded successfully!")
print("Number of rows and columns:", df.shape)

display(df.head())
```

```python

from pathlib import Path

print("Current project location:")
print(Path.cwd())

print("\nFiles and folders in project:")
for item in Path.cwd().iterdir():
    print(item.name)

print("\nFiles inside data folder:")
data_folder = Path("data")

if data_folder.exists():
    for item in data_folder.iterdir():
        print(repr(item.name))
else:
    print("data folder not found in current location")
```

```python

# 1. Check dataset size
print("Dataset Shape:", df.shape)

# 2. Check column names and data types
print("\n--- Column Information ---")
df.info()

# 3. Check missing values
print("\n--- Missing Values ---")
print(df.isnull().sum())

# 4. Check duplicate rows
print("\n--- Duplicate Rows ---")
print("Duplicate rows:", df.duplicated().sum())

# 5. Get summary statistics
print("\n--- Numerical Summary ---")
display(df.describe())
```

```python

# Check loan status categories
print("--- Loan Status Counts ---")
display(df["LoanStatus"].value_counts(dropna=False))

# Check missing dates by loan status
print("\n--- Loan Status vs Paid In Full Date ---")
display(pd.crosstab(
    df["LoanStatus"],
    df["PaidInFullDate"].isna(),
    margins=True
))

print("\n--- Loan Status vs Charge-Off Date ---")
display(pd.crosstab(
    df["LoanStatus"],
    df["ChargeOffDate"].isna(),
    margins=True
))

# Check the few missing business details
print("\n--- Missing Business Details ---")
display(df[df["BusinessType"].isna() | df["BusinessAge"].isna()])
```

```python

# Create a separate copy for cleaning
df_clean = df.copy()

# Convert date columns to proper datetime format
date_columns = [
    "ApprovalDate",
    "PaidInFullDate",
    "ChargeOffDate"
]

for col in date_columns:
    df_clean[col] = pd.to_datetime(
        df_clean[col],
        format="%d-%m-%Y",
        errors="coerce"
    )

# Check date conversion
print("--- Date Data Types ---")
print(df_clean[date_columns].dtypes)

# Check missing business details
print("\n--- Missing Values Before Handling ---")
print(df_clean[["BusinessType", "BusinessAge"]].isnull().sum())

# Fill missing categorical details with "Unknown"
df_clean["BusinessType"] = df_clean["BusinessType"].fillna("Unknown")
df_clean["BusinessAge"] = df_clean["BusinessAge"].fillna("Unknown")

# Verify cleaning
print("\n--- Missing Values After Handling ---")
print(df_clean[["BusinessType", "BusinessAge"]].isnull().sum())

print("\nCleaned dataset shape:", df_clean.shape)
```

```python

# Count loans by status
loan_status_counts = (
    df_clean["LoanStatus"]
    .value_counts()
    .rename_axis("LoanStatus")
    .reset_index(name="LoanCount")
)

# Calculate percentage of total loans
loan_status_counts["Percentage"] = (
    loan_status_counts["LoanCount"] / len(df_clean) * 100
).round(2)

print("--- Loan Status Summary ---")
display(loan_status_counts)

# Calculate total charge-off amount
total_charge_off = df_clean["GrossChargeOffAmount"].sum()

print(
    "\nTotal Gross Charge-Off Amount: $",
    f"{total_charge_off:,.2f}"
)

# Count loans with a positive charge-off amount
charge_off_loans = (
    df_clean["GrossChargeOffAmount"] > 0
).sum()

print("Loans with positive charge-off amount:", charge_off_loans)
```

```python

# Analyze loan approvals by financial year
yearly_analysis = (
    df_clean.groupby("ApprovalFY")
    .agg(
        TotalLoans=("ApprovalFY", "size"),
        TotalApprovalAmount=("GrossApproval", "sum"),
        AverageLoanAmount=("GrossApproval", "mean"),
        TotalJobsSupported=("JobsSupported", "sum")
    )
    .reset_index()
)

# Format amounts for readability
yearly_analysis["AverageLoanAmount"] = (
    yearly_analysis["AverageLoanAmount"].round(2)
)

yearly_analysis["TotalApprovalAmount"] = (
    yearly_analysis["TotalApprovalAmount"].round(2)
)

print("--- Yearly Loan Approval Analysis ---")
display(yearly_analysis)
```

```python

print("Number of yearly records:", len(yearly_analysis))
print(yearly_analysis.to_string(index=False))
```

```python

# Compare processing methods
method_analysis = (
    df_clean.groupby("ProcessingMethod")
    .agg(
        TotalLoans=("ProcessingMethod", "size"),
        TotalApprovalAmount=("GrossApproval", "sum"),
        ChargeOffLoans=("LoanStatus", lambda s: (s == "CHGOFF").sum()),
        ChargeOffAmount=("GrossChargeOffAmount", "sum")
    )
    .reset_index()
)

method_analysis["ChargeOffPercentage"] = (
    method_analysis["ChargeOffLoans"]
    / method_analysis["TotalLoans"] * 100
).round(2)

print("--- Processing Method Analysis ---")
display(method_analysis)
```

```python

# Analyze loan status by business age
business_age_analysis = (
    df_clean.groupby("BusinessAge")
    .agg(
        TotalLoans=("LoanStatus", "size"),
        ChargeOffLoans=(
            "LoanStatus",
            lambda s: (s == "CHGOFF").sum()
        ),
        TotalApprovalAmount=("GrossApproval", "sum"),
        ChargeOffAmount=("GrossChargeOffAmount", "sum")
    )
    .reset_index()
)

business_age_analysis["ChargeOffPercentage"] = (
    business_age_analysis["ChargeOffLoans"]
    / business_age_analysis["TotalLoans"] * 100
).round(2)

print("--- Business Age Analysis ---")
display(
    business_age_analysis.sort_values(
        "ChargeOffPercentage",
        ascending=False
    )
)
```

```python

print(
    business_age_analysis[
        [
            "BusinessAge",
            "TotalLoans",
            "ChargeOffLoans",
            "TotalApprovalAmount",
            "ChargeOffAmount",
            "ChargeOffPercentage"
        ]
    ].to_string(index=False)
)
```

```python

from pathlib import Path

# Create output folder
output_folder = Path("data/processed")
output_folder.mkdir(parents=True, exist_ok=True)

# Save cleaned dataset
output_file = output_folder / "sba_loan_cleaned.csv"

df_clean.to_csv(output_file, index=False)

print("Cleaned dataset saved successfully!")
print("File location:", output_file.resolve())
print("Rows and columns:", df_clean.shape)

# Verify the saved file
df_check = pd.read_csv(output_file)

print("Saved file verification:", df_check.shape)
```
